**Ejercicio 6**

Aplicar round, floor, ceil y fix sobre [1.2, 2.8, -3.4, -5.7] y explicar las diferencias.

Funciones: np.round(), np.floor(), np.ceil(), np.fix()

```python
# Ejemplo ilustrativo (con otros valores)
v_ejemplo = np.array([0.5, -0.5, 2.5])
print("Vector ejemplo:", v_ejemplo)
print("round ejemplo:", np.round(v_ejemplo))
print("floor ejemplo:", np.floor(v_ejemplo))
print("ceil ejemplo:", np.ceil(v_ejemplo))
print("fix ejemplo:", np.fix(v_ejemplo))
```

```python
#@title Resultado
v = np.array([1.2, 2.8, -3.4, -5.7])

print("Vector:", v)
print("round:", np.round(v))
print("floor:", np.floor(v))
print("ceil:", np.ceil(v))
print("fix:", np.fix(v))

print("""
Explicacion:
- round redondea al entero mas cercano.
- floor siempre redondea hacia abajo (hacia -infinito).
- ceil siempre redondea hacia arriba (hacia +infinito).
- fix trunca la parte decimal, redondeando siempre hacia cero.
Por eso floor y fix coinciden en los positivos, pero difieren en los negativos:
floor(-3.4) = -4, mientras que fix(-3.4) = -3 (trunca hacia cero).
""")
```

**Ejercicio 9**

Calcular el determinante y la matriz inversa de una matriz. Verificar que A^-1 * A = I.

Funciones: np.linalg.det(), np.linalg.inv()

```python
# Ejemplo ilustrativo (con otros valores)
M_ejemplo = np.array([[3.0, 1.0], [2.0, 1.0]])
print("Determinante ejemplo:", np.linalg.det(M_ejemplo))
print("Inversa ejemplo:\n", np.linalg.inv(M_ejemplo))
```

```python
#@title Resultado
A = np.array([[4, 7], [2, 6]], dtype=float)

det_A = np.linalg.det(A)
inv_A = np.linalg.inv(A)
verificacion = inv_A @ A

print("A:\n", A)
print("Determinante:", det_A)
print("Inversa:\n", inv_A)
print("A^-1 @ A (deberia ser la identidad):\n", np.round(verificacion, 8))
```

**Ejercicio 10**

Calcular autovalores y autovectores de una matriz e interpretar el resultado.

Funciones: np.linalg.eig()

```python
# Ejemplo ilustrativo (con otros valores)
M_ejemplo = np.array([[1.0, 2.0], [2.0, 1.0]])
val_ejemplo, vec_ejemplo = np.linalg.eig(M_ejemplo)
print("Autovalores ejemplo:", val_ejemplo)
print("Autovectores ejemplo:\n", vec_ejemplo)
```

```python
#@title Resultado
A = np.array([[2, 0], [0, 3]], dtype=float)
autovalores, autovectores = np.linalg.eig(A)

print("A:\n", A)
print("Autovalores:", autovalores)
print("Autovectores:\n", autovectores)
print("""
Interpretacion: los autovalores (2 y 3) indican el factor de escala que aplica
la matriz A sobre sus autovectores asociados (los versores canonicos, en este
caso, porque A es diagonal). Cada autovector define una direccion que la
transformacion A solo estira o encoge, sin rotarla.
""")
```

**Bonus 4 - Distancias entre usuarios**

Dados tres usuarios representados mediante vectores, calcular la distancia Euclidiana y la similitud coseno entre cada par, y determinar cuales son los usuarios mas parecidos y mas diferentes.

Funciones: np.dot(), np.linalg.norm()

```python
# Ejemplo ilustrativo (con otros valores)
u_ejemplo = np.array([1, 0])
v_ejemplo = np.array([0, 1])
print("Distancia euclidiana ejemplo:", np.linalg.norm(u_ejemplo - v_ejemplo))
print("Similitud coseno ejemplo:", np.dot(u_ejemplo, v_ejemplo) / (np.linalg.norm(u_ejemplo) * np.linalg.norm(v_ejemplo)))
```

```python
#@title Resultado
usuario_1 = np.array([5, 3, 0, 1])
usuario_2 = np.array([4, 3, 0, 1])
usuario_3 = np.array([1, 0, 5, 4])

usuarios = {"Usuario 1": usuario_1, "Usuario 2": usuario_2, "Usuario 3": usuario_3}

def distancia_euclidiana(a, b):
    return np.linalg.norm(a - b)

def similitud_coseno(a, b):
    return np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b))

pares = [("Usuario 1", "Usuario 2"), ("Usuario 1", "Usuario 3"), ("Usuario 2", "Usuario 3")]
for nombre_a, nombre_b in pares:
    a, b = usuarios[nombre_a], usuarios[nombre_b]
    print(f"{nombre_a} vs {nombre_b} -> Euclidiana: {distancia_euclidiana(a, b):.3f} | "
          f"Coseno: {similitud_coseno(a, b):.3f}")

print("""
Los usuarios 1 y 2 son los mas parecidos: tienen la menor distancia Euclidiana
y la mayor similitud coseno (cercana a 1). Los usuarios 1 y 3 (y 2 y 3) son
los mas diferentes, con mayor distancia y menor similitud coseno.
""")
```

Graficar la funcion logistica (sigmoide) y luego calcular su inversa (funcion logit), verificando algunos valores. sigma(x) = 1 / (1 + e^(-x))

Funciones: np.exp(), np.log(), plt.plot()

```python
# Ejemplo ilustrativo (con otros valores)
def sigmoide_ejemplo(x):
    return 1 / (1 + np.exp(-x))

print("Sigmoide ejemplo en [-1, 0, 1]:", sigmoide_ejemplo(np.array([-1, 0, 1])))
```

```python
#@title Resultado
def sigmoide(x):
    return 1 / (1 + np.exp(-x))

def logit(p):
    return np.log(p / (1 - p))

x = np.linspace(-10, 10, 200)
y = sigmoide(x)

plt.figure(figsize=(6, 4))
plt.plot(x, y, color="darkorange")
plt.title("Funcion logistica (sigmoide)")
plt.xlabel("x")
plt.ylabel("sigma(x)")
plt.grid(True, alpha=0.3)
plt.show()

valores_prueba = np.array([-2, 0, 1, 3])
recuperados = logit(sigmoide(valores_prueba))

print("Valores originales:", valores_prueba)
print("sigmoide(x):", sigmoide(valores_prueba))
print("logit(sigmoide(x)) (deberia recuperar x):", np.round(recuperados, 8))
```
